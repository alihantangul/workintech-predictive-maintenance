```python
import pandas as pd

# 1. Dosya yollarını tanımlama
file_1 = '/content/drive/MyDrive/Workintech Data Analytics Bitirme Projesi/1_Asset_93707_MDEMAIG180-22B_2026-04.xlsx'
file_2 = '/content/drive/MyDrive/Workintech Data Analytics Bitirme Projesi/2_Asset_93707_MDEMAIG180-22B_2026-04.xlsx'

print("Veriler 'Health Score Data Model' klasöründen okunuyor ve işleniyor. Lütfen bekleyin...")

# 2. Excel dosyalarını okuma (ABB portalı 2. satırda başlıkları verdiği için header=1 dedim)
df1_raw = pd.read_excel(file_1, header=1)
df2_raw = pd.read_excel(file_2, header=1)

# 3. İlk dosyadan (Operasyon ve Genel Durum) verileri çekme
df1 = pd.DataFrame()
df1['Timestamp'] = pd.to_datetime(df1_raw.iloc[:, 0])
df1['Overall_Vib_mm_s'] = pd.to_numeric(df1_raw.iloc[:, 2], errors='coerce')
df1['Speed_rpm'] = pd.to_numeric(df1_raw.iloc[:, 5], errors='coerce')
df1['Skin_Temp_C'] = pd.to_numeric(df1_raw.iloc[:, 8], errors='coerce')
df1['Frequency_Hz'] = pd.to_numeric(df1_raw.iloc[:, 11], errors='coerce')
df1['Output_Power_kW'] = pd.to_numeric(df1_raw.iloc[:, 14], errors='coerce')
df1['Pk_Pk_Tangential_g'] = pd.to_numeric(df1_raw.iloc[:, 17], errors='coerce')

# 4. İkinci dosyadan (Yönsel İvme ve Titreşim) verileri çekme
df2 = pd.DataFrame()
df2['Timestamp'] = pd.to_datetime(df2_raw.iloc[:, 0])
df2['Vib_Radial_mm_s'] = pd.to_numeric(df2_raw.iloc[:, 2], errors='coerce')
df2['Vib_Tangential_mm_s'] = pd.to_numeric(df2_raw.iloc[:, 5], errors='coerce')
df2['Vib_Axial_mm_s'] = pd.to_numeric(df2_raw.iloc[:, 8], errors='coerce')
df2['Acc_RMS_Axial_g'] = pd.to_numeric(df2_raw.iloc[:, 11], errors='coerce')
df2['Acc_RMS_Tangential_g'] = pd.to_numeric(df2_raw.iloc[:, 14], errors='coerce')
df2['Acc_RMS_Radial_g'] = pd.to_numeric(df2_raw.iloc[:, 17], errors='coerce')

# 5. İki veri setini 'Timestamp' anahtarı üzerinden birleştirme
df_merged = pd.merge(df1, df2, on='Timestamp', how='inner')

# 6. VERİ TEMİZLİĞİ: Sadece yüke bindiği anları (Speed > 0) alıp, hatalı okumaları (NaN) temizleme
df_clean = df_merged[df_merged['Speed_rpm'] > 0].dropna().reset_index(drop=True)

# 7. İşlem sonucunu raporlama
print("-" * 50)
print("VERİ HAZIRLIĞI TAMAMLANDI!")
print("-" * 50)
print(f"Toplam Birleştirilen Satır Sayısı (Motor Kapalıyken Dahil) : {len(df_merged)}")
print(f"Model İçin Temizlenen Satır Sayısı (Sadece Yükteyken)      : {len(df_clean)}")
print("-" * 50)

# İlk 5 satırın önizlemesi
display(df_clean.head())
```
